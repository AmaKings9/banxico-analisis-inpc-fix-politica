# Documentación del proyecto de análisis de la relación entre la inflación (INPC), el tipo de cambio (USD/MXN) y la tasa objetivo de Banxico entre 2018 y 2019

## Objetivo del proyecto
Entender la relación entre tres de los índices más importantes dentro de la macroeconomía nacional para entender cómo la política monetaria responde ante la inflación y cómo se relaciona con el tipo de cambio.

## Preguntas a resolver
1. ¿Cómo ha evolucionado la inflación (INPC), el tipo de cambio (FIX) y la tasa objetivo desde el año 2018 hasta 2025?
2. ¿La tasa objetivo responde a los cambios en la inflación?
3. ¿Los cambios en la tasa objetivo explican los movimientos del tipo de cambio?

## Conceptos relevantes
- **Indice Nacional de Precios al Consumidor (INPC):** es un indicador estadístico que mide la variación promedio generalizada de los precios de una canasta básica dentro de un periodo establecido. Se calcula de forma quincenal y mensual, es a partir de este valor que se determina la inflación.
- **Inflación:** Es el aumento prolongado de los precios de bienes y servicios en un periodo de tiempo, lo que provoca la disminución del valor real del dinero.
- **Tipo de Cambio (FIX, USD/MXN):** Referencia del valor de la moneda nacional para transacciones que involucren el intercambio de divisas.
- **Tasa objetivo:** Es la tasa de interés a pagar que establece el Banco de México para los bancos acorde a los recursos monetarios que soliciten. A partir de esta tasa, el Banco de México pretende influir en las tasas de interés de créditos y depósitos que ofrecen los bancos privados, lo que impactará en el consumo, producción e inflación.

### 1. Extracción de datos
Los datos a analizar se obtuvieron de la API pública del Banco de México a través de una extracción manual en consola local (esto debido a que la plataforma no se encontraba redireccionando correctamente a la ruta de la nueva API al ejecutar una petición tipo get) y se almacenó la respuesta en archivos formato json. Posteriormente, se rescataron las columnas correspondientes a la fecha y a cada uno de los índices, los cuales se almacenaron en un dataframe de dos columnas por cada índice económico en formato CSV.

*Este proceso se encuentra codificado en la ruta src/extraer_data.py*

### 2. Limpieza de datos
Los datos de los tres dataframes, correspondientes a los índices económicos previamente listados, se unieron en un solo dataframe utilizando la función *merge out*, tomando como índice la columna fecha. Recordemos que la función *merge out* une todos los registros de dos o más dataframes basandose en una o más columnas índices comunes para formar un solo dataframe.

Además, el análisis previo a la limpieza mostró una diferencia en número de registros no nulos entre el FIX, INPC y la tasa objetivo por lo que para mantener la continuidad de los datos en el nuevo dataframe, se aplicó la función *forward fill* sobre la columna tasa objetivo.
Recordemos que la función *forward fill* se utiliza para reemplazar a los valores nulos dentro del dataframe a partir de propagar el último valor válido conocido hacía los campos nulos.

Finalmente, se almacenó el nuevo dataframe limpio en un archivo con formato CSV.

*Este proceso se encuentra codificado en la ruta src/limpieza_data.py*

### 3. Preparación de datos
Este proceso se centró en calcular las variaciones mensuales y anuales, estas últimas solo para los índices FIX y INPC, de cada uno de los índices a partir de modificar el periodo del dataframe de días a meses, obteniendo el último dato de cada mes como valor representantivo. 

Finalmente se almacenó el nuevo dataframe, con las nuevas columnas de las variaciones mensuales y anuales integradas, en un archivo nuevo con formato CSV.

*Este proceso se encuentra codificado en la ruta src/preparar_data.py*

### 4. Exploración de datos
En la siguientes gráficas de series de tiempo se muestra el comportamiento a través del tiempo de los índices variación anual del INPC (Inflación anual), FIX mensual y Tasa Objetivo mensual desde el año 2018 hasta el primero de agosto del año 2025.

#### 4.1 Serie de tiempo de la inflación anual (2018-2025)
**(a) ¿Cómo ha evolucionado la inflación entre 2019 y 2025?**

La inflación en México durante este periodo se puede dividir en tres etapas. La primera etapa aborda desde el periodo 2019-2020 donde presenta un comportamiento usual a pesar de presentarse una ligera disminución de alrededor de 1%,, ya que sus valores se encuentran entre el 3-4%. La segunda etapa presenta lo que se le conoce como un shock inflacionario que se describe como el aumento repentino y generalizado de los precios de los bienes y servicios de una economía, lo que provoca una disminución en la oferta; entre 2021 y 2022, la inflación crece hasta niveles cercanos al 9%. Finalmente, en la tercera fase comienza la desaceleración gradual, la cual comienza desde 2023 hasta el año 2024, donde alcanza poco más del 4%, y luego la inflación regresa a oscilar en el rango 3-4%.

**(b) ¿Hay periodos donde la inflación cambia de forma destacada?**

El cambio más destacado ocurre entre 2021 y 2022, el cual coincide con la disrupción monetaria global dentro del periodo post-pandemia, donde los precios internacionales aumentaros a niveles fuera de lo común. Afortunadamente este comportamiento finaliza con la deflación sostenida entre 2023 y 2024.

#### 4.2 Serie de tiempo del Tipo de Cambio (2018-2025)
**(a) ¿Cómo ha evolucionado el tipo de cambio entre 2018 y 2025?**

El tipo de cambio tambíen muestra cuatro etapas diferentes durante el periodo de 2018 a 2025.

La primera etapa que abarca de 2018 a inicios de 2020 muestra una estabilidad relativa con un rango acotado de 18 a poco más de 20 pesos por dólar; un periodo sin cambios abruptos que refleja un entorno económico internacional estable.

Por otro lado, en la segunda etapa de 2020 a mediados de ese mismo año, el tipo de cambio se deprecia rápidamente, alcanzando niveles cercanos a los 24 pesos por dólar; este periodo coincide con el inicio oficial de la pandemia por COVID-19, que generó incertidumbre global, salida de capitales de mercados emergentes y caída en los precios internacionales del petróleo.

En contraste, el periodo que comienza en la segunda mitad del 2020 hasta inicios del 2021, muestra un aumento continuo de la apreciación de la moneda nacional, que lo lleva a niveles de entre 17 y 18 pesos por dólar, el cual se mantendría hasta finales de 2022. Este comportamiento se asocia con la recuperación económica global.

Finalmente, desde 2023 hacía finales del 2024 se presentó una apreciación de casi tres pesos, la cual duró poco ya que a finales de este episodio se observa un regreso a niveles cercanos a 19-20 pesos por dólar.

**(b) ¿Hay periodos donde el tipo de cambio cambia de forma destacada?**

Se identifican tres principales periodos con variaciones significativas:

El movimiento más notable se encuentra dentro de la primera mitad del 2020, donde el peso se deprecia rápidamente desde niveles cercanos a 19 hasta casi 24 pesos por dólar. Este comportamiento está directamente relacionado con la crisis sanitaria global. Le sigue la apreciación sostenida desde mediados de 2020 a mitad de 2023, la cual está relacionada con la recuperación económica. Y, por último, de 2024 a 25 cuando el tipo de cambio vuelve a retomar sus valores dentro del rango habitual; este comportamiento puede atribuirse a las elecciones electorales que ocurrieron en el país dentro de ese periodo.

#### 4.3 Serie de tiempo de la tasa de objetivo (2018-2025)
**(a) ¿Cómo ha evolucionado la tasa objetivo entre 2018 y 2025?**

En este periodo la tasa objetivo sigue una estructura similar a la inflación, ya que presenta tes fases claramente diferenciadas.

Desde 2018 hasta 2019 la tasa se mantiene entre poco más de 7% y 8.25%, es decir, es decir, se mantiene con un crecimiento estable. No es hasta finales de 2019 que la tasa objetivo baja drásticamente hasta llegar al 4% en 2021; este comportamiento coincide con la contracción económica mundial ocasionado por la crisis del COVID-19. A mediados de 2021, el Banco de México decide incrementar la tasa de forma constante, alcanzando un máximo histórico de 11.25% en 2023. Finalmente comienza una disminución controlada de la tasa hasta llegar al 8% aproximadamente en 2025.

**(b) ¿Hay periodos donde la tasa objetivo cambia de forma destacada?**
El periodo de 2020-2021 muestra la disminución abrupta de la tasa, el cual demuestra que el Banco de México requería la necesidad de estímulo monetario, que por el contrario a mediados de 2021 a 2023, este mismo índice sufrió un incremento acelerado para responder al choque inflacionario global. Ambos periodos demuestran un cambio repentino de la tasa objetivo

#### 4.4 Series de tiempo Inflación VS Tasa Objetivo (2018-2025)

**(a) ¿La tasa objetivo responde a cambios en la inflación?**

La gráfica muestra simultáneamente la variación anual del INPC (inflación) y la tasa objetivo del Banco de México para el periodo 2018–2025. La comparación revela una relación clara entre ambos indicadores; sin embargo, la respuesta de la tasa objetivo no es inmediata ante los cambios en la inflación, lo que refleja el rezago natural con el que la política monetaria actúa sobre la economía. Esto también indica que el valor de la tasa objetivo no depende únicamente de la inflación, sino de un conjunto más amplio de factores económicos internos y externos.

Entre 2020 y 2021, la inflación se mantuvo relativamente baja (entre 3% y 4%). Durante este periodo, la tasa objetivo también se redujo de forma significativa, pasando de 8% a niveles cercanos al 4%. Esto sugiere una respuesta de similar por parte de la tasa objetivo ante la inflación.

A partir de finales de 2021 e inicios de 2022, la inflación comenzó a incrementarse fuertemente, alcanzando niveles cercanos al 8.7% en 2022–2023. La tasa objetivo reaccionó con un aumento sostenido y acelerado, subiendo desde 4% hasta 11.25%, lo que coincide con la estrategia antiinflacionaria del Banco de México.

Después de 2023, la inflación inició una tendencia a la baja. Del mismo modo, durante 2024–2025 la tasa objetivo muestra una reducción gradual, reflejando nuevamente una respuesta alineada con la trayectoria descendente de la inflación.

#### 4.5 Scatter del Tipo de cambio y la Tasa Objetivo (2018-2025)

**(a) ¿Los cambios en la tasa objetivo explican los movimientos del tipo de cambio?**

El diagrama de dispersión muestra la relación entre el tipo de cambio FIX (pesos por dólar) y la tasa objetivo del Banco de México durante el periodo 2018–2025 donde cada punto representa un mes.

Los puntos no siguen un patrón ascendente o descendente consistente. Se observan valores de tipo de cambio similares asociados tanto a tasas altas como bajas. Por ejemplo, cuando la tasa objetivo está entre 7% y 9%, el tipo de cambio varía desde valores cercanos a 18 hasta casi 22 pesos por dólar. Esto indica que la tasa objetivo no es el índice principal que describe los movimientos del tipo de cambio y sugiere que el tipo de cambio responde también a otros factores macroeconómicos y financieros.
