# Documentación del proyecto de análisis de la relación entre la inflación (INPC), el tipo de cambio (USD/MXN) y la tasa objetivo de Banxico entre 2018 y 2019

## Objetivo del proyecto
Entender la relación entre tres de los índices más importantes dentro de la macroeconomía nacional para entender cómo la política monetaria responde ante la inflación y cómo se relaciona con el tipo de cambio.

## Preguntas a resolver
1. ¿Cómo ha evolucionado la inflación (INPC), el tipo de cambio (FIX) y la tasa objetivo desde el año 2018 hasta 2025?
2. ¿La tasa objetivo responde a los cambios en la inflación?
3. ¿Los cambios en la tasa objetivo explican los movimientos del tipo de cambio?

## Conceptos relevantes
- **Indice Nacional de Precios al Consumidor (INPC):** es un indicador estadístico que mide la variación promedio generalizada de los precios de una canasta básica dentro de un periodo establecido. Se calcula de forma quincenal y mensual, es a partir de este valor que se determina la inflación.
- **Inflación:** Es el aumento prolongado de los precios de bienes y servicios en un periodo de tiempo, lo que provoca la disminución del valor real del dinero.
- **Tipo de Cambio (FIX, USD/MXN):** Referencia del valor de la moneda nacional para transacciones que involucren el intercambio de divisas.
- **Tasa objetivo:** Es la tasa de interés a pagar que establece el Banco de México para los bancos acorde a los recursos monetarios que soliciten. A partir de esta tasa, el Banco de México pretende influir en las tasas de interés de créditos y depósitos que ofrecen los bancos privados, lo que impactará en el consumo, producción e inflación.

### 1. Extracción de datos
Los datos a analizar se obtuvieron de la API pública del Banco de México a través de una extracción manual en consola local (esto debido a que la plataforma no se encontraba redireccionando correctamente a la ruta de la nueva API al ejecutar una petición tipo get) y se almacenó la respuesta en archivos formato json. Posteriormente, se rescataron las columnas correspondientes a la fecha y a cada uno de los índices, los cuales se almacenaron en un dataframe de dos columnas por cada índice económico en formato CSV.

*Este proceso se encuentra codificado en la ruta src/extraer_data.py*

### 2. Limpieza de datos
Los datos de los tres dataframes, correspondientes a los índices económicos previamente listados, se unieron en un solo dataframe utilizando la función *merge out*, tomando como índice la columna fecha. Recordemos que la función *merge out* une todos los registros de dos o más dataframes basandose en una o más columnas índices comunes para formar un solo dataframe.

Además, el análisis previo a la limpieza mostró una diferencia en número de registros no nulos entre el FIX, INPC y la tasa objetivo por lo que para mantener la continuidad de los datos en el nuevo dataframe, se aplicó la función *forward fill* sobre la columna tasa objetivo.
Recordemos que la función *forward fill* se utiliza para reemplazar a los valores nulos dentro del dataframe a partir de propagar el último valor válido conocido hacía los campos nulos.

Finalmente, se almacenó el nuevo dataframe limpio en un archivo con formato CSV.

*Este proceso se encuentra codificado en la ruta src/limpieza_data.py*

### 3. Preparación de datos
Este proceso se centró en calcular las variaciones mensuales y anuales, estas últimas solo para los índices FIX y INPC, de cada uno de los índices a partir de modificar el periodo del dataframe de días a meses, obteniendo el último dato de cada mes como valor representantivo. 

Finalmente se almacenó el nuevo dataframe, con las nuevas columnas de las variaciones mensuales y anuales integradas, en un archivo nuevo con formato CSV.

*Este proceso se encuentra codificado en la ruta src/preparar_data.py*
