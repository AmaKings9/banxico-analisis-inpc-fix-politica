## Exploración de datos posterior a su limpieza

```python
macro_m_df = pd.read_csv("../data_clean/macro_mensual_mx_2018_2025.csv", parse_dates=["fecha"])
macro_m_df.set_index("fecha")
macro_m_df.info()
```

```python
macro_m_df.head(15)
```

```python
macro_m_df[["inpc_var_a"]].tail()
```

```python
import matplotlib.pyplot as plt
import datetime as dt
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(macro_m_df["fecha"], macro_m_df["inpc_var_a"], color="red")
ax.set_xlim(left=dt.datetime(2018, 1, 1))
ax.set_title("Inflación anual (variación anual del INPC en %)")
ax.set_xlabel("Fecha (2018-2025)")
ax.set_ylabel("Inflación anual (%)")
plt.savefig("../figuras/serie_inflacion_2018_2025.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(macro_m_df["fecha"], macro_m_df["fix"], color="blue")
ax.set_title("Tipo de Cambio FIX (mensual)")
ax.set_xlabel("Fecha (2018-2025)")
ax.set_ylabel("FIX (pesos por dólar)")
plt.savefig("../figuras/serie_fix_2018_2025.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(macro_m_df["fecha"], macro_m_df["tasa_obj"], color="green")
ax.set_title("Tasa objetivo (mensual en %)")
ax.set_xlabel("Fecha (2018-2025)")
ax.set_ylabel("Tasa Objetivo (%)")
plt.savefig("../figuras/serie_tasa_obj_2018_2025.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8, 6))
ax.plot(macro_m_df["fecha"], macro_m_df["inpc_var_a"], color="cornflowerblue", label="Inflación anual")
ax.plot(macro_m_df["fecha"], macro_m_df["tasa_obj"], color="navy", label="Tasa Objetivo", linestyle="--")
ax.set_xlim(left=dt.datetime(2018, 1, 1))
ax.set_ylim(0, 12)
ax.set_title("Inflación anual (variación anual del INPC) VS Tasa Objetivo (mensual)")
ax.set_xlabel("Fecha (2018-2025)")
ax.set_ylabel("Porcentaje (%)")
ax.grid(color="lightgray", linewidth=0.5)
ax.legend()
plt.savefig("../figuras/inflacion_vs_tasa_obj_2018_2025.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fig, ax_scatter = plt.subplots(figsize=(8, 6))
ax_scatter.scatter(macro_m_df["fix"], macro_m_df["tasa_obj"])
ax_scatter.set_title("Correlación entre el Tipo de Cambio FIX y la Tasa Objetivo (mensual)")
ax_scatter.set_xlabel("Tipo de Cambio FIX (pesos por dólar)")
ax_scatter.set_ylabel("Tasa Objetivo (%)")
plt.savefig("../figuras/scatter_fix_tasa_obj_2018_2025.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
fix, ax_line = plt.subplots(figsize=(8, 6))
ax_line.plot(macro_m_df["fecha"], macro_m_df["fix"], color="blue")
ax_line.plot(macro_m_df["fecha"], macro_m_df["tasa_obj"], color="pink")
ax_line.set_title("Tipo de Cambio FIX VS Tasa Objetivo")
ax_line.set_xlabel("Fecha (2018-2025)")
ax_line.set_ylabel("Valor")
ax.legend()
```
